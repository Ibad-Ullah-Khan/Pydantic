## __Pydantic:__
Pydantic is a Python library used for data validation and data parsing. It ensures that data follows the expected structure and types.
### __Data Validation__
Checking whether the received data follows the defined rules.

__Example:__ Checking whether age is an integer.
### __Data Parsing__
Converting received data into the required format or structure.

__Example:__ Converting "22" (string) into 22 (integer), when supported by Pydantic's conversion ru

```python
from pydantic import BaseModel #BaseModel is a class of pydantic library

# Creating pydantic model: We use BaseModel to create pydantic model
class Patient(BaseModel): # This is pydantic model

    age: int
    name: str

def insert_patient_data(patient: Patient):
    print(patient.name)
    print(patient.age)
    print('Data inserted')

patient_data = {'name': 'ibad', 'age': 21}
patient1 = Patient(**patient_data) #This is a pydantic object

insert_patient_data(patient1)
```

### __Why Pydantic__
We use Pydantic to:
1. Data Validation: Check whether received data follows the defined rules and data types.
2. Data Parsing: Convert received data into the required Python types.
3. Data Structure: Ensure data follows a predefined structure.
4. Error Handling: Identify invalid data and provide validation errors.

### __More in Detail__

```python
# More details

from pydantic import BaseModel
from typing import List, Dict

class Patient(BaseModel):

    name: str
    age: int
    weight: float
    married: bool
    allergies: List[str]
    contact_details: Dict[str, str]


def update_patient_data(patient: Patient):

    print(patient.name)
    print(patient.age)
    print(patient.allergies)
    print(patient.married)
    print('updated')

patient_info = {'name':'ibad','age': '30', 'weight': 75.2, 'married': False, 'allergies':['pollen', 'dust'],'contact_details':{'phone':'2353462'}}

patient1 = Patient(**patient_info)
update_patient_data(patient1)

# list: Accepts a list without specifying item types.
# List[str]: Ensures all items inside the list are strings.
# Dict[str, str]: Ensures dictionary keys and values are strings.
# That's why we improting the List and Dict class
```

### __Optional Cases__

```python
# More details

from pydantic import BaseModel 
from typing import List, Dict, Optional

class Patient(BaseModel):

    name: str 
    age: int
    weight: float
    married: bool = False# You can also give a default value like you given below in optional case
    allergies: Optional[List[str]] = None 
    contact_details: Optional[Dict[str, str]] = None


def update_patient_data(patient: Patient):

    print(patient.name)
    print(patient.age)
    print(patient.allergies)
    print(patient.married)
    print(patient.contact_details)
    print('updated')

patient_info = {'name':'ibad','age': '30', 'weight': 75.2, 'married': True, 'contact_details':{'phone':'2353462'}}

patient1 = Patient(**patient_info)
update_patient_data(patient1)

# Optional: Allows a field to accept its specified type or None.
# Example: Optional[str] means a string or None.
```

### __Special Format__

### __Constrains Case__

```python

from pydantic import BaseModel, EmailStr, AnyUrl, Field #used to add extra validation rules and constraints to model fields.
from typing import List, Dict, Optional

class Patient(BaseModel):

    name: str = Field(max_length=60)
    email: EmailStr 
    linked_in: AnyUrl
    age: int = Field(gt=0, lt=120) # If input is not in this rangle, system will through error
    weight: float = Field(gt=0)
    married: bool
    allergies: List[str]
    contact_details: Dict[str, str]


def update_patient_data(patient: Patient):

    print(patient.name)
    print(patient.age)
    print(patient.allergies)
    print(patient.married)
    print('updated')

patient_info = {
    'name': 'ibad',
    'linked_in': 'http://linkedin.com/11224',
    'email': 'abc@gmail.com',
    'age': '30',
    'weight': 75.2,
    'married': False,
    'allergies': ['pollen', 'dust'],
    'contact_details': {'phone': '2353462'}
}

patient1 = Patient(**patient_info)

update_patient_data(patient1)
```

### __Description Case__

```python

from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict, Optional, Annotated # Annotated add extra information or metadata to a type without changing its original type.

class Patient(BaseModel):

    name: Annotated[str, Field(max_length=50, title='Name of the patient', description='Add the name of the patient in less then 50 characters', examples=['ibad', 'Salman'])]
    email: EmailStr 
    linked_in: AnyUrl
    age: int = Field(gt=0, lt=120)
    weight: float = Field(gt=0)
    married: Annotated[bool, Field(default=False, description='Is the patient is married or not')]
    allergies: List[str]
    contact_details: Dict[str, str]


def update_patient_data(patient: Patient):

    print(patient.name)
    print(patient.age)
    print(patient.allergies)
    print(patient.married)
    print('updated')

patient_info = {
    'name': 'ibad',
    'linked_in': 'http://linkedin.com/11224',
    'email': 'abc@gmail.com',
    'age': '30',
    'weight': 75.2,
    'married': False,
    'allergies': ['pollen', 'dust'],
    'contact_details': {'phone': '2353462'}
}

patient1 = Patient(**patient_info)

update_patient_data(patient1)
```

```python
# But sometime by default converstion can make problem for you, to avoid it, you can do
from pydantic import BaseModel, Field
from typing import Annotated
class Patient(BaseModel): 

    name: str
    age: Annotated[int, Field(gt=0, strict=True)]

def insert_patient_data(patient: Patient):
    print(patient.name)
    print(patient.age)
    print('Data inserted')

patient_data = {'name': 'ibad', 'age': 21} # Now string eg: '21' will not work
patient1 = Patient(**patient_data)

insert_patient_data(patient1) 
```
