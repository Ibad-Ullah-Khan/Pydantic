### __Field Validator__
A field validator is used to define custom validation rules for a specific field in a Pydantic model.

Custom validation means creating your own rules to check whether data meets your specific requirements

For example, you want to ensure that a patient's name contains only letters.



```python
from pydantic import BaseModel, field_validator

class Patient(BaseModel):

    name: str
    age: int

    @field_validator('name')
    @classmethod
    def validate_name(cls, value):
        if not value.isalpha():
            raise ValueError('Name must contain only letters')
        return value
```

### __mode parameter__
The mode parameter determines when your custom validator runs during Pydantic's validation process.

There are three commonly used modes:
1. after
2. before

#### __1: 2. After mode__
mode='after' is the default mode.

It runs after Pydantic has performed its standard type validation and conversion.

```python
from pydantic import BaseModel, field_validator

class Patient(BaseModel):

    age: int

    @field_validator('age', mode='after')
    @classmethod
    def validate_age(cls, value):
        if value < 18:
            raise ValueError('Age must be 18 or above')
        return value

patient = Patient(age='21')
```

#### __Before mode__
mode='before' runs before Pydantic's standard type validation.

```python
from pydantic import BaseModel, field_validator

class Patient(BaseModel):

    age: int

    @field_validator('age', mode='before')
    @classmethod
    def validate_age(cls, value):
        print(type(value))
        return value

patient = Patient(age='21')

# We return value from every field Validator because Pydantic needs the validated value to store in the model field.
```
